```python
import pandas as pd
```

## First dataset (from [muaban.net](https://muaban.net/bat-dong-san/ban-nha))

```python
df_1 = pd.read_csv('../data/interim/muaban_net.csv').drop(columns='Unnamed: 0')
df_1.columns
```

## Second dataset (from [batdongsan.com.vn](https://batdongsan.com.vn/nha-dat-ban))

```python
df_2 = pd.read_csv('../data/interim/batdongsan_com_vn(1).csv').drop(columns='Unnamed: 0')
df_2.columns
```

## Merge raw datasets


```python
df = pd.concat([df_1, df_2], axis=0).reset_index().drop(columns=['index', 'verified', 'date_of_posting', 'dimension_2'])
df.info()
```

```python
df.to_csv('../data/interim/full.csv')
```
