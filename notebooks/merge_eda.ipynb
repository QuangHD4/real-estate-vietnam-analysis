```python
import pandas as pd
```

```python
df_1 = pd.read_csv('../data/interim/batdongsan_com_vn(1).csv')
df_2 = pd.read_csv('../data/interim/batdongsan_com_vn(2).csv')
df_3 = pd.read_csv('../data/interim/muaban_net.csv')
```

### Merge datasets

```python
cols_in_common = [
    'price', 'area', 'n_bedrooms', 'n_bathrooms', 'property_type', 'legal_docs', 
    'city_province', 'district', 'address', 'facing_direction', 'front_width'
]
df = pd.concat([df_1[cols_in_common], df_2[cols_in_common], df_3[cols_in_common]], axis=0)
```

```python
df.info()
```

```python
df.city_province.unique()
```

`city_province` needs a little standardization for 'Bà Rịa Vũng Tàu'

```python
df.loc[df.city_province == 'Bà Rịa - Vũng Tàu', 'city_province'] = 'Bà Rịa Vũng Tàu'
```

## Checkpoint: Check and Save merged dataset
